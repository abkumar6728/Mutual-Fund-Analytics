```python
import requests
import pandas as pd

# AMFI Scheme Code
scheme_code = "120503"

# API URL
url = f"https://api.mfapi.in/mf/{scheme_code}"

# Fetch data
response = requests.get(url)

# Check if request was successful
print("Status Code:", response.status_code)

if response.status_code == 200:
    data = response.json()

    print("\nScheme Name:")
    print(data["meta"]["scheme_name"])

    print("\nLatest NAV:")
    print(data["data"][0])

else:
    print("Failed to fetch data.")
```

```python
import pandas as pd

# Convert the latest NAV record into a DataFrame
latest_nav = pd.DataFrame([data["data"][0]])

print("Latest NAV:")
display(latest_nav)
```

```python
from pathlib import Path

# Create the processed data folder if it doesn't exist
output_folder = Path("../data/processed")
output_folder.mkdir(parents=True, exist_ok=True)

# Save the latest NAV to a CSV file
latest_nav.to_csv(output_folder / "live_nav.csv", index=False)

print("✅ Live NAV saved to data/processed/live_nav.csv")
```
